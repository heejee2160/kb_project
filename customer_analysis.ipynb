```python
import getpass
import oracledb

password = getpass.getpass("Oracle 비밀번호: ")

conn = oracledb.connect(
    user="SMALLBIZ",
    password=password,
    dsn="192.168.1.108:1521/ORCLPDB1",
    tcp_connect_timeout=10
)

print("Oracle DB 연결 성공!")
```

```python
import pandas as pd

customer_df = pd.read_sql("""
    SELECT *
    FROM CUSTOMER
""", conn)

print("CUSTOMER 데이터 크기:", customer_df.shape)
print("고객 수:", customer_df["CUSTOMER_ID"].nunique())

customer_df.head()
```

```python
product_df = pd.read_sql("""
    SELECT *
    FROM FINANCIAL_PRODUCT
""", conn)

print("FINANCIAL_PRODUCT 데이터 크기:", product_df.shape)
print("금융상품 가입 건수:", len(product_df))

product_df.head()
```

```python
analysis_df = product_df.merge(
    customer_df,
    on="CUSTOMER_ID",
    how="left"
)

print("분석 데이터 크기:", analysis_df.shape)
print("고객 수:", analysis_df["CUSTOMER_ID"].nunique())
print("결측 고객 수:", analysis_df["REGION_CODE"].isna().sum())

analysis_df.head()
```

```python
industry_df = pd.read_sql("""
    SELECT *
    FROM INDUSTRY
""", conn)

region_df = pd.read_sql("""
    SELECT *
    FROM REGION
""", conn)

print("INDUSTRY 컬럼:", industry_df.columns.tolist())
print("REGION 컬럼:", region_df.columns.tolist())

display(industry_df.head())
display(region_df.head())
```

```python
analysis_df = (
    analysis_df
    .merge(
        industry_df,
        on="INDUSTRY_ID",
        how="left"
    )
    .merge(
        region_df,
        on="REGION_CODE",
        how="left"
    )
)

print("최종 분석 데이터 크기:", analysis_df.shape)
print("업종명 결측:", analysis_df["INDUSTRY_NAME"].isna().sum())
print("지역명 결측:", analysis_df["REGION_NAME"].isna().sum())

analysis_df.head()
```

```python
# Pandas 고객 금융 분석

## 1. 상품별 이용 비중
금융상품별 가입 건수와 전체 가입 건 대비 이용 비중을 분석한다.

## 2. 업종별 평균 가입금액
업종별 금융상품 평균 가입금액을 비교한다.

## 3. 대출 보유 여부별 고객 비교
대출 보유 고객과 미보유 고객의 고객 수, 평균 매출,
평균 매출증감률을 비교한다.
```

```python
# 1. 상품별 이용 비중

product_ratio = (
    analysis_df["PRODUCT_NAME"]
    .value_counts()
    .reset_index()
)

product_ratio.columns = ["상품명", "이용건수"]

product_ratio["이용비중(%)"] = (
    product_ratio["이용건수"]
    / product_ratio["이용건수"].sum()
    * 100
).round(2)

product_ratio
```

```python
# 2. 업종별 평균 가입금액

industry_amount = (
    analysis_df
    .groupby("INDUSTRY_NAME")["AMOUNT"]
    .mean()
    .reset_index()
)

industry_amount.columns = ["업종", "평균가입금액"]

industry_amount["평균가입금액"] = (
    industry_amount["평균가입금액"]
    .round(0)
    .astype(int)
)

industry_amount = industry_amount.sort_values(
    "평균가입금액",
    ascending=False
)

industry_amount
```

```python
# 3. 대출 보유 여부별 고객 비교

loan_compare = (
    customer_df
    .groupby("HAS_LOAN")
    .agg(
        고객수=("CUSTOMER_ID", "count"),
        평균매출=("VIRTUAL_SALES_AMOUNT", "mean"),
        평균매출증감률=("SALES_GROWTH_RATE", "mean")
    )
    .reset_index()
)

loan_compare["HAS_LOAN"] = loan_compare["HAS_LOAN"].map({
    0: "대출 미보유",
    1: "대출 보유"
})

loan_compare["평균매출"] = loan_compare["평균매출"].round(0)
loan_compare["평균매출증감률"] = loan_compare["평균매출증감률"].round(2)

loan_compare
```

```python
# 분석 결과 저장

product_ratio.to_csv(
    "product_ratio.csv",
    index=False,
    encoding="utf-8-sig"
)

industry_amount.to_csv(
    "industry_amount.csv",
    index=False,
    encoding="utf-8-sig"
)

loan_compare.to_csv(
    "loan_compare.csv",
    index=False,
    encoding="utf-8-sig"
)

print("분석 결과 저장 완료")
```
